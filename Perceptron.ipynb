**PERCEPTRON**

A perceptron is the simplest form of an artificial neural network and the basic building block of deep learning, invented by Frank Rosenblatt in 1957 to mimic how a biological neuron makes decisions.







**How a Perceptron Works?**

A perceptron takes a set of numerical inputs, processes them through a mathematical formula, and produces a single binary output (0 or 1).
1. Inputs (\(x_1, x_2, \dots, x_n\)): The raw features or measurable data points fed into the model.
2. Weights (\(w_1, w_2, \dots, w_n\)): Values that show the importance of each input; higher weights mean that input has a stronger influence on the decision.
3. Bias (b): A constant added to the weighted sum to shift the decision boundary up or down.
4. Weighted Sum (z): The model calculates \(z = (w_1 x_1 + w_2 x_2 + \dots + w_n x_n) + b\).
5. Activation Function: A threshold function (like a step function) that turns the weighted sum into a final binary class label, usually 1 or 0

In 2d, perceptron acts as a line.




And  in 3d,perceptron acts as a plane.







And in 4d, perceptron acts as a hyperplane.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = df.drop(columns=['Gender','Stream','Hostel'])
```

```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.linear_model import Perceptron
c=Perceptron()
```

```python
x=X[['Age','CGPA']]
```

```python
c.fit(x.values, y.values)
```

```python
c.coef_
```

```python
c.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(x.values,y.values,clf=c)

plt.show()
```
